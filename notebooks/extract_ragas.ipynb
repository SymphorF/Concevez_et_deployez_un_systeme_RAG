```python
import os
import pandas as pd
from datasets import Dataset
import asyncio
import nest_asyncio
import traceback
from typing import List, Dict
from langchain_mistralai.chat_models import ChatMistralAI
from langchain_mistralai.embeddings import MistralAIEmbeddings
from ragas import evaluate
from dotenv import load_dotenv
from ragas.metrics import (
    faithfulness,
    answer_relevancy,
    context_precision,
    context_recall,
)

nest_asyncio.apply()

# --- CORRECTION : Extraction du contenu textuel des contextes ---
def extract_contexts_content(contexts_dicts):
    """Extrait le contenu textuel des dictionnaires de contexte"""
    contexts_text = []
    for context_dict in contexts_dicts:
        if isinstance(context_dict, dict) and 'content' in context_dict:
            contexts_text.append(context_dict['content'])
        elif isinstance(context_dict, str):
            contexts_text.append(context_dict)
        else:
            # Fallback : convertir en string
            contexts_text.append(str(context_dict))
    return contexts_text

# Préparation des données avec contextes corrigés
evaluation_data = {
    "question": questions_test,
    "answer": answers,
    "contexts": [extract_contexts_content(ctx) for ctx in placeholder_contexts],  # 🔹 CORRIGÉ
    "ground_truth": ground_truths
}

evaluation_dataset = Dataset.from_dict(evaluation_data)
print("Dataset d'évaluation prêt.")
print(f"Exemple de contexte: {evaluation_dataset[0]['contexts'][0][:100]}...")  # Vérification

# --- Configuration et Exécution de l'Évaluation ---
load_dotenv(dotenv_path=os.path.join("..", ".env"))

MISTRAL_API_KEY = os.getenv("MISTRAL_API_KEY")
if not MISTRAL_API_KEY:
    print("⚠️ AVERTISSEMENT : Clé API Mistral non trouvée ou non définie.")
else:
    print("Clé API Mistral chargée avec succès !")

try:
    # 1. Initialisation du LLM et des Embeddings
    print("Initialisation LLM et Embeddings Mistral...")
    mistral_llm = ChatMistralAI(mistral_api_key=MISTRAL_API_KEY, model="mistral-large-latest", temperature=0.1)
    mistral_embeddings = MistralAIEmbeddings(mistral_api_key=MISTRAL_API_KEY)
    print("LLM et Embeddings initialisés.")

    # 2. Définition des métriques
    metrics_to_evaluate = [
        faithfulness,
        answer_relevancy,
        #context_precision,
        #context_recall,
    ]
    print(f"Métriques sélectionnées: {[m.name for m in metrics_to_evaluate]}")

    # 3. Lancement de l'évaluation Ragas
    print("\nLancement de l'évaluation Ragas (peut prendre du temps)...")
    results = evaluate(
        dataset=evaluation_dataset,
        metrics=metrics_to_evaluate,
        llm=mistral_llm,
        embeddings=mistral_embeddings
    )
    print("\n--- Évaluation Ragas terminée ---")

    # 4. Affichage des résultats
    print("\n--- Résultats de l'évaluation (DataFrame) ---")
    results_df = results.to_pandas()
    pd.set_option('display.max_rows', None)
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.max_colwidth', 150)
    print(results_df)

    # 5. Scores moyens
    print("\n--- Scores Moyens (sur tout le dataset) ---")
    average_scores = results_df.mean(numeric_only=True)
    print(average_scores)

except Exception as e:
    print(f"\n❌ ERREUR lors de l'initialisation ou de l'évaluation Ragas : {e}")
    print("\nTraceback:")
    traceback.print_exc()
```

- des dates inventées

- des horaires ou prix non présents dans le contexte

- des lieux approximatifs

- une synthèse un peu trop imaginative

- des transitions ou explications non présentes dans la source

**📌 Exemple d’interprétation**

Si le contexte dit :

“Un atelier de cuisine aura lieu le 14 janvier à Paris.”

Et que ta réponse dit :

“L’atelier de cuisine du 14 janvier aura lieu à Paris, et il proposera des recettes italiennes.”

→ Si le contexte ne mentionne pas les recettes italiennes, alors cette petite invention fait baisser le score à ~0.7.

**Ce score est-il bon ?**

Oui, 0.71 est correct, surtout dans un RAG early stage.
Cela montre que le système est fonctionnel, mais :

👉 on peut encore réduire les hallucinations.
👉 viser 0.8–0.9 pour un système très fiable.

📈 Recommandation d'amélioration

✔️ 1. Améliorer le prompt RAG

Par exemple :

“Réponds uniquement en utilisant les informations du contexte.
Ne fais aucune supposition.”

✔️ 2. Tester d'autres LLM plus robuste comme GPT-4.1


## II- Answer_relevancy

**Un answer_relevancy = 0.6 (sur une échelle de 0 à 1) signifie ceci :**

### ✅ Interprétation simple

#### Le RAG répond globalement à la question.

*0.60 = 60 % de pertinence*

La réponse est liée à la question, mais :

- il manque des éléments importants, ou

- une partie du contenu est hors-sujet, ou

- la réponse est trop vague, ou

- elle mélange des infos utiles avec du bruit.

🔍 Comment RAGAS juge answer_relevancy ?

La métrique mesure :

✔️ Pertinence directe

La réponse traite-t-elle réellement la question ?

✔️ Exhaustivité minimale

La réponse couvre-t-elle les éléments attendus ?

✔️ Cohérence

La réponse ne part-elle pas trop loin dans des détails inutiles ?

**⚠️ Même si le contexte est bon, RAGAS regarde si la réponse en particulier correspond vraiment à la question.**

**Donc 0.6 = la réponse est correcte, mais peu être amélioré pour rendre le système RAG plus robuste.**


#### 🧩 Pourquoi tu peux avoir 0.6 ?

Voici les causes possinbles :

🔸 1. RAG répond trop large ou trop général

→ Pas assez spécifique à la question.

🔸 2. Il reformule sans réellement répondre

→ Un peu “blabla”.

🔸 3. Partie de la réponse pas liée à la question

→ Du bruit.

🔸 4. Éléments attendus absents

→ Exemple : la question demande des événements ce week-end, mais la réponse donne une activité mensuelle.

🔸 5. Contexte peu utile ou mal utilisé

→ Le modèle comble avec de l’imagination.



#### 🎯 Que faire pour améliorer un score de 0.6 ?

✔️ 1. Améliorer les chunks

- Plus courts

- Plus centrés sur un seul événement

- Éviter les paragraphes longs et polyvalents

✔️ 2. Ajouter des métadonnées (ville, dates, catégories)

→ Cela aide l’indexation et donc la pertinence.


📌 En résumé

Un answer_relevancy de 0.6 veut dire que ton RAG :

comprend partiellement la question,

répond en partie correctement,

mais produit une réponse trop vague, ou incomplète.
